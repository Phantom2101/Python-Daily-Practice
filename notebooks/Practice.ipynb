```python
# Q1. Given sales = {"Mon": 200, "Tue": 350, "Wed": 150, "Thu": 400}, print the day with the highest sales and the total sales.

sales = {"Mon": 200, "Tue": 350, "Wed": 150, "Thu": 400}

highest_day = max(sales, key = sales.get)
total_sales = sum(sales.values())

print(highest_day)
print(total_sales)
```

```python
# Q2. Given students = [{"name": "Ravi", "marks": 72}, {"name": "Anu", "marks": 88}, {"name": "Sam", "marks": 65}], print only the names of students with marks above 70.

students = [{"name": "Ravi", "marks": 72}, {"name": "Anu", "marks": 88}, {"name": "Sam", "marks": 65}]

for i in students:
  if i["marks"] > 70 :
    print(i["name"])
```

```python
#Q4. Write grade(marks) that returns "A" for 85+, "B" for 70-84, "C" for 50-69, and "F" otherwise. Then use it to print each student's grade from the list in question 2.
# FUNCTON PRACTICE

def stud_grad(marks):
  if marks >= 85:
    return '--Grade A'
  elif 70 <= marks < 84:
    return '--Grade B'
  elif 50 <= marks < 69:
    return '--Grade C'
  else:
    return '--Grade F'

students = [{"name": "Ravi", "marks": 72}, {"name": "Anu", "marks": 88}, {"name": "Sam", "marks": 65}, {"name": "Kashish", "marks": 36}]
for i in students:
  print(i['name'],stud_grad(i["marks"]))



```

```python
# Q5. 5. Write count_words(text) that returns a dictionary with each word and its frequency. Make it case-insensitive.
#FUNCTION PRACTICE

def count_words(text):
  word = text.lower().split()
  count = {}

  for i in word:
    if i in count:
      count[i] += 1
    else:
      count[i] = 1
  return count

text = "Apple banana Apple orange banana apple"

print(count_words(text))

```

```python
# Q7. Using a dictionary comprehension, turn ["data", "sql", "python"] into {"data": 4, "sql": 3, "python": 6} (word to length).

list1 = ["data", "sql", "python"]

compr = {i:len(i) for i in list1 }
print(compr)
```

```python
# Q8. Practice

# Given this list of numbers:

# numbers = [1, 2, 3, 4, 5, 6]

# Using a dictionary comprehension, create a dictionary where:

# The number is the key
# The square of the number is the value

numbers = [1, 2, 3, 4, 5, 6]

dictionary_comprehension = {i : i*i for i in numbers }
print(dictionary_comprehension)


```

```python
# Q9. Write remove_duplicates(items) that returns a list with duplicates removed but keeps the original order. Try it without using set() first.

numbers = [6, 5, 4, 3, 2, 11, 2, 3, 4, 5, 6, 1, 7, 7, 4, 2, 3, 8 ]
#numbers.reverse()
# print(numbers)
def remove_duplicates(items):
  result = []

  for i in items:
    if i not in result:
      result.append(i)

  return result

print(remove_duplicates(numbers))
```
