```python
# Armstrong number

def len_of_integer(n):
  count = 0
  while n > 0:
      count += 1
      n //= 10
  return count

def arm_num(n):
  length = len_of_integer(n)
  r = 0
  temp = n
  for i in range(1, length +1):
      p = temp % 10
      temp //= 10
      q = p ** length
      r = r + q

  # print(r)
  if r == n:
    return "Armstrong number"
  else:
    return "Not armstrong number"

n = int(input("Enter no.= "))
print(arm_num(n))
```

```python
# Find the missing number in an unsorted list (by specifying start and count)

def sum_of_num(n):
  total = 0
  for i in n:
    total += i
  return total

def missing_num(n, start_num, expected_count):
  if not n:
    return None

  # Since we know the starting number and total count of numbers,
  # we can precisely find the ending number of the consecutive range.
  end_num = start_num + expected_count - 1

  # Calculate the expected sum of this range
  expected_sum = (expected_count * (start_num + end_num)) // 2

  # The missing number is simply the expected sum minus the actual sum of the elements present
  return expected_sum - sum_of_num(n)

# User Inputs
start_no = int(input("Enter the starting number of the sequence: "))
total_count = int(input("Enter how many numbers should be in the list (including the missing one): "))
numbers = [int(x) for x in input("Enter list numbers: ").split()]

print("Missing number:", missing_num(numbers, start_no, total_count))
```

```python
# Q28: Find the Majority Element

# Difficulty: Medium

# Given a list of integers, find the number that appears more than half the total number of elements.

# Example 1
# Input:  [1, 2, 2, 2, 3, 2, 4]
# Output: 2
number = [int(x) for x in input("Enter numbers: ").split()]

def majority_elem(n):
  seen = []
  freq = []
  for i in n:
    if i not in seen:
      count = 0
      for j in n:
        if j == i:
          count += 1
      seen.append(i)
      freq.append(count)
  count_of_max_occuring_element = max(freq)

  if count_of_max_occuring_element > len(n) / 2:
    for i in range(len(freq)):
        if freq[i] == count_of_max_occuring_element:
          return seen[i]
  return None
print(majority_elem(number))

```

```python
# Q29: Find the Leaders in a List
#Problem

# Given a list of integers, find all leaders in the list.

# An element is a leader if it is greater than or equal to every element to its right. The last element is always a leader.

# Example 1
# Input:  [16, 17, 4, 3, 5, 2]
# Output: [17, 5, 2]
numbers = [int(x) for x in input("enter numbers:").split()]

def leaders(n):
  Leaders = []
  for i in range(len(n)):
    current_num = n[i]
    is_Leader = True

    for j in range(i+1, len(n)):
      if current_num <= n[j]:
        is_Leader = False
        break
    if is_Leader:
      Leaders.append(current_num)

  return Leaders

print(leaders(numbers))
```

```python
# Q32: Merge Two Sorted Lists
# Medium

# Lists · Traversal · Sorting logic

# Problem

# Given two lists that are already sorted in ascending order, merge them into one sorted list.

raw_numbers_1 = input("Enter numbers in list 1 (in ascending order only): ").split()
raw_numbers_2 = input("Enter numbers in list 2 (in ascending order only): ").split()

list_1 = [int(x) for x in raw_numbers_1] if raw_numbers_1 else []
list_2 = [int(x) for x in raw_numbers_2] if raw_numbers_2 else []

def sorted_merged_list(m, n):
  i = 0
  j = 0
  final_list = []

  while i < len(m) and j < len(n):
    if m[i] <= n[j]:
      final_list.append(m[i])
      i += 1
    else:
      final_list.append(n[j])
      j +=1

  while i < len(m):
    final_list.append(m[i])
    i += 1

  while j < len(n):
    final_list.append(n[j])
    j +=1

  return final_list

print(sorted_merged_list(list_1, list_2))
```

```python
# Q35: Find the Longest Increasing Contiguous Sequence
# Medium · Lists · Index Tracking

# Given a list of integers, find the length of the longest contiguous sequence that is strictly increasing.

# A contiguous sequence means the elements must be next to each other in the original list.

# Example 1

# Input: [1, 2, 3, 1, 2, 5, 6]

# Output: 4

# Explanation: The longest increasing contiguous sequence is [1, 2, 5, 6], with length 4.

# Example 2

# Input: [5, 4, 3, 2]

# Output: 1
raw_numbers = input("Enter numbers in list : ").split()

numbers = [int(x) for x in raw_numbers] if raw_numbers else []

def contigous_len(n):
  if len(n) == 0:
    return 0

  current_len = 1
  max_len = 1

  for i in range(1, len(n)):
    if n[i] > n[i - 1]:
      current_len += 1
    else:
      current_len = 1
    if current_len > max_len:
      max_len = current_len

  return max_len

print(contigous_len(numbers))
```

```python
# Q37: Majority Element — Efficient Version
# Medium · Lists · Algorithm Optimization

# You've already solved the majority-element problem using frequency counting. Now let's improve it.

# Problem: Given a list, find the element that appears more than half the length of the list. Return None if no such element exists.

# Examples:

# [3, 3, 4, 3, 2, 3, 3] → 3

# [1, 2, 3, 2, 1] → None

# [5, 5, 5, 2, 3] → 5

# New challenge: Try to solve it in O(n) time without building a frequency list or repeatedly counting each element.

# Hint: Think about keeping a candidate and a count. When you encounter the same value, increase the count; when you encounter a different value, decrease it.

# There's one important detail: after choosing a candidate, you must verify that it really appears more than half the time.

raw_numbers = input("Enter numbers in list : ").split()

numbers = [int(x) for x in raw_numbers] if raw_numbers else []

def majority_Elem(n):
  candidate = None
  count = 0

  for i in n:
    if count == 0:
      candidate = i
      count = 1
    elif i == candidate:
      count += 1
    else:
      count -= 1

  actual_count = 0
  for i in n:
    if i == candidate:
      actual_count += 1

  if actual_count > len(n) // 2:
      return candidate

  return None

print(majority_Elem(numbers))


```

```python
# Q40: Binary Search

# Medium · Lists · Searching Algorithms

# Problem: Given a list of integers sorted in ascending order and a target number, find the index of the target. If the target is not present, return -1.

# Example 1:

# Input: numbers = [2, 5, 8, 12, 16, 23, 38], target = 16

# Output: 4

# Example 2:

# Input: numbers = [2, 5, 8, 12, 16, 23, 38], target = 7

# Output: -1

# Rules:

# The input list is already sorted in ascending order.

# Use a loop; don't use index() or in to search.

# Don't use linear search.

# Return the index of the target, or -1 if it isn't present.

# Hint: Instead of checking every element, examine the middle element. If the target is smaller, search the left half;
#       if it's larger, search the right half.

raw_num = input("enter numbers:").split()
tgt = input("enter target:").strip()
numbers = [int(x) for x in raw_num] if raw_num else []
target = int(tgt) if tgt else 0

def binary_search(n, t):
   low = 0
   high = len(n) - 1

   while low <= high:
    mid = (low + high) // 2

    if n[mid] == t:
      return mid
    elif t < n[mid]:
      high = mid - 1
    else:
      low = mid + 1


   return -1


print(binary_search(numbers, target))



```

```python
# Q41: Bubble Sort

# Easy · Lists · Sorting Algorithms

# Problem

# Given a list of integers, sort the list in ascending order using the Bubble Sort algorithm.

# Bubble Sort works by comparing neighboring elements and swapping them if they are in the wrong order. Repeat this process until the list is sorted.

# Example 1

# Input:

# numbers = [5, 2, 8, 1, 3]

# Output:

# [1, 2, 3, 5, 8]
raw_num = input("enter numbers:").split()
numbers = [int(x) for x in raw_num] if raw_num else []

def bubble_sort(n):
  for i in range(len(n)):
    for j in range(0, len(n) - i - 1):
      if n[j] > n[j + 1]:
        n[j], n[j + 1] = n[j + 1], n[j]
  return n

print(bubble_sort(numbers))


```

```python
# Q42: Selection Sort

# Easy · Lists · Sorting Algorithms

# Problem

# Given a list of integers, sort it in ascending order using the Selection Sort algorithm.

# Selection Sort repeatedly finds the smallest element in the unsorted portion of the list and swaps it with the first element of that portion.

raw_num = input("enter numbers:").split()
numbers = [int(x) for x in raw_num] if raw_num else []

def selection_Sort(n):
  for i in range(0, len(n) - 1):
    min = i
    for j in range(i+1, len(n)):
      if n[j] < n[min]:
        min = j
    if min != i:
      n[i], n[min] = n[min], n[i]
  return n

print(selection_Sort(numbers))
```

```python
# Q43: Insertion Sort

# Easy · Lists · Sorting Algorithms

# Problem

# Given a list of integers, sort the list in ascending order using the Insertion Sort algorithm.

# Insertion Sort builds a sorted portion of the list one element at a time. It takes the next element and inserts it into the correct position among the elements already sorted.

raw_num = input("enter numbers:").split()
numbers = [int(x) for x in raw_num] if raw_num else []

def insertion_Sort(n):
  for i in range(1, len(n)):
    temp = n[i]
    j = i - 1

    while j >= 0 and n[j] > temp:
      n[j + 1] = n [j]
      j -= 1

    n[j + 1] = temp

  return n
print(insertion_Sort(numbers))
```
